```python
#Min and Max Numbers 
lst = [12, 3, 4, 5, 6, 7]

big = lst[0]
small = lst[0]

for i in lst:
    if i > big:
        big = i

    if i < small:
        small = i

print("Biggest =", big)
print("Smallest =", small)
```

```python
#Pascal's Triangle 
n = int(input("Enter rows: "))
for i in range(n):
    num = 1
    for s in range(n - i):
        print(" ", end="")
    for j in range(i + 1):
        print(num, end=" ")
        num = num * (i - j) // (j + 1)
    print()
```
