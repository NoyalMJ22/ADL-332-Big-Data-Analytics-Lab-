```python
#kmeans 
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt 

iris= load_iris()
x=iris.data

model=KMeans(n_clusters=3)
model.fit(x)

y=model.predict(x)

plt.scatter(x[:,0],x[:,1],c=y)

plt.xlabel("SepalLeng")
plt.ylabel("SepalWidth")

plt.show()

```
